```python
import pandas as pd
import math

# Read CSV
data = pd.read_csv("play_tennis.csv")

target = "Play"

pos = data[data[target] == "Yes"].to_dict("records")
neg = data[data[target] == "No"].to_dict("records")

rules = []


def cover(rule, examples):
    return [
        x for x in examples
        if all(x[k] == v for k, v in rule)
    ]


while pos:

    rule = []

    while cover(rule, neg):

        best = None
        best_gain = -999

        for feature in data.columns[:-1]:

            for value in data[feature].unique():

                literal = (feature, value)

                if literal in rule:
                    continue

                new_rule = rule + [literal]

                p0 = len(cover(rule, pos))
                n0 = len(cover(rule, neg))

                p1 = len(cover(new_rule, pos))
                n1 = len(cover(new_rule, neg))

                if p1 == 0:
                    continue

                gain = p1 * (
                    math.log2(p1 / (p1 + n1))
                    - math.log2(p0 / (p0 + n0))
                )

                if gain > best_gain:
                    best_gain = gain
                    best = literal

        if best is None:
            break

        rule.append(best)

    rules.append(rule)

    covered = cover(rule, pos)
    pos = [x for x in pos if x not in covered]


# Print rules
print("\nLearned FOIL Rules:\n")

for rule in rules:
    conditions = " AND ".join(
        f"{k}={v}" for k, v in rule
    )

    print(f"IF {conditions} THEN Play=Yes")
```
