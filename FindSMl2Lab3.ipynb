```python
import pandas as pd
import numpy as np
df = pd.read_csv("diabetes_risk.csv")
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
# Initial hypothesis
h = np.array(["Ø"] * X.shape[1])
print("Initial Hypothesis:", h)
print("-" * 60)
```

```python
for i, (x, target) in enumerate(zip(X, y), start=1):
    print(f"Iteration {i}")
    print("Training Example:", x)
    print("Target:", target)
    if target == "Yes":
        h = np.where(
            h == "Ø",
            x,
            np.where(h == x, h, "?")
        )
        print("Positive example → Hypothesis updated")
    else:
        print("Negative example → Hypothesis unchanged")
    print("Hypothesis:", h)
    print("-" * 60)

print("Final Hypothesis:", h)
```
